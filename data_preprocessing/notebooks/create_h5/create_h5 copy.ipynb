```python
from pkl_h5 import PKL2H5
import logging
ASTRO_OBJ_INPUT_PATH = '/home/magdalena/pipeline/data_preprocessing/data/AO_PKL/2023_out/'
ASTRO_OBJ_INPUT_DIRS = glob.glob('/home/magdalena/pipeline/data_preprocessing/data/OID_PARQUETS/*')
print(ASTRO_OBJ_INPUT_DIRS) #sanity check!
CROSS_CONT_FILE = '/home/magdalena/pipeline/data_preprocessing/data/FF_DATASET/partitions.parquet'
OUTPUT_DIR = './data/H5_FILES/'
```

```python
logger = logging.getLogger()
log_file_name = 'multiple_h5.log'
logging.basicConfig(filename=log_file_name,
                    encoding='utf-8',
                    level=logging.DEBUG,
                    filemode='w',  # This will overwrite the log file
                    format = "[%(asctime)s] [%(levelname)s]: %(message)s", datefmt="%H:%M:%S"
                    )
```

```python
h5_creator = PKL2H5(ASTRO_OBJ_INPUT_PATH,
                '/home/magdalena/pipeline/data_preprocessing/data/test/',
                CROSS_CONT_FILE,
                'test.h5',
                logger=logger,
                purge_output_directory=False
                )
 
h5_creator.create_dataset(train_val_split=0.2,
                            k_folds=5,
                        generate_features_qt= True,
                        generate_metadata_qt=True)
```

```python
for year_dir in ASTRO_OBJ_INPUT_DIRS:
    match = re.search(r"/(\d{4})_", year_dir)
    year = match.group(1)
    output_dir = os.path.join(OUTPUT_DIR,str(year))
    output_dataset_name = 'clean_'+str(year)+'.h5'
    h5_creator = PKL2H5(year_dir,
                output_dir,
                CROSS_CONT_FILE,
                output_dataset_name,
                logger=logger,
                purge_output_directory=False
                )
    
    
    h5_creator.create_dataset(train_val_split=0.2,
                              k_folds=5,
                            generate_features_qt= True,
                            generate_metadata_qt=True)
```
