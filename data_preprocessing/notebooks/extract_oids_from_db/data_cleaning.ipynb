```python
import pandas as pd
import glob
```

```python
core_path = '/home/magdalena/pipeline/data_preprocessing/data/OID_PARQUETS/'
parquets = glob.glob(f'{core_path}*.parquet')
print(parquets)
```

```python
dfs = []
for file in parquets:
    df = pd.read_parquet(f'{file}')
    #print(df.columns)
    #df.sort_values('ndet')
    df = df[df['ndet']>6]
    dfs.append(pd.DataFrame(df))
```
