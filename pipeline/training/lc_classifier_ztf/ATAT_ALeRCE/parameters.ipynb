```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
params_lc = np.array([14.62,21.09,27.55,
                    53.82,78.53,104.26,
                    205.95,305.54,405.12,])

emb_size = np.array([32,64,128]*3
                    )
depth = np.array([1,2,3]*3)
scores_lc = np.array([0.4760,.5011,0.4904,
                      0.4843,0.5010,0.4916,
                      0.4942,0.4947,.4988])


params_tab = np.array([18.46,24.93,31.39,
                       49.22,99.65,74.43,
                       147.58,247.17,346.75
                       ])


scores_tab = np.array([.6865,.6817,0.6592,
                       0.6988,0.6816 ,0.6138,
                       0.6977,0.6921,0.6077])
```

```python
import pandas as pd

df_lc = pd.DataFrame(
    {"params": params_lc, "scores": scores_lc, "params": params_tab, 'emb_size':emb_size, 'depth':depth}
)
df_tab = pd.DataFrame(
    {"params": params_tab, "scores": scores_tab, "params": params_tab,'emb_size':emb_size, 'depth':depth}
)
df_lc['type'] = 'lc'
df_tab['type'] = 'tab'

df_final = pd.concat([df_lc, df_tab])
```

```python
df_final
```

```python
sns.scatterplot(data = df_final, x = 'params', y = 'scores', hue = 'type')
plt.xlabel('Parameter Count')
plt.ylabel('F1-score (macro)')
plt.grid('on')
```
